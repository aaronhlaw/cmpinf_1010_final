```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('MonthlyCriminalActivity.csv', dtype={18: str})

# Count rows per neighborhood, biggest first
counts = df.groupby('Neighborhood').size().sort_values(ascending=False)

# Print the x neighborhoods with the fewest entries
numSafestNeighborhoods = 10
print("Top ", numSafestNeighborhoods, " safest neighborhoods (fewest entries):")
print(counts.sort_values(ascending=True).head(numSafestNeighborhoods))

# Plot
counts.plot(kind='bar', figsize=(12, 6), color='seagreen')
plt.title('Number of Entries per Neighborhood')
plt.xlabel('Neighborhood')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right', fontsize=6)
plt.tight_layout()
plt.show()
```
