```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv('MonthlyCriminalActivity.csv')

# Plotting a basic vertical bar chart
df.plot.bar(x='Neighborhood', y='Sales', color='skyblue', legend=False)

# Customizing titles and labels
plt.title('Criminal Activity in a Neighborhood')
plt.xlabel('Neighborhood')
plt.ylabel('Crime Amount')
plt.xticks(rotation=0)  # Keeps the x-axis labels horizontal
plt.show()
```
